Bài 16: Nhâ ̣p vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma ̣
trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải

```python
n, m = map(int, input().split())

a = []
for i in range(n):
    row = list(map(int, input().split()))
    a.append(row)

for row in a:
    for x in row:
        print(f"{x:>4}", end="")
    print()
```

Bài 17: Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và
đường chéo phụ của mảng

```python
n = int(input())

a = []
for i in range(n):
    row = list(map(int, input().split()))
    a.append(row)

print("Duong cheo chinh:")
for i in range(n):
    print(a[i][i], end=" ")

print("\nDuong cheo phu:")
for i in range(n):
    print(a[i][n - 1 - i], end=" ")
```

Bài 19: Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất
trong list

```python
def vi_tri_max(a):
    max_pos = 0

    for i in range(1, len(a)):
        if a[i] > a[max_pos]:
            max_pos = i

    return max_pos

a = list(map(int, input().split()))
print(vi_tri_max(a))
```

Bài 20: Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và
trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về
-1

```python
def tim_k(a, k):
    for i in range(len(a)):
        if a[i] == k:
            return i
    return -1

a = list(map(int, input().split()))
k = int(input())

print(tim_k(a, k))
```

Bài 21: Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ
chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện

```python
s = input().split()

d = {}

for i in range(len(s)):
    if s[i] not in d:
        d[s[i]] = []

    d[s[i]].append(i)

for key in d:
    d[key] = tuple(d[key])

print(d)
```

Bài 23: VIết hàm xóa đi các số <x có trong mảng

```python
def xoa_nho_hon_x(a, x):
    return [i for i in a if i >= x]

a = list(map(int, input().split()))
x = int(input())

print(xoa_nho_hon_x(a, x))
```

Bài 24: Viêết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí
trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 2 3 2 4 9

```python
def day_0_ve_dau(a):
    zero = []
    other = []

    for x in a:
        if x == 0:
            zero.append(x)
        else:
            other.append(x)

    return zero + other

a = list(map(int, input().split()))
print(day_0_ve_dau(a))
```

Bài 26: Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2
phần tử đầu và cuối

```python
a = list(map(int, input().split()))

tong = 0

for i in range(1, len(a) - 1):
    if a[i] > a[i - 1] and a[i] > a[i + 1]:
        tong += a[i]

print(tong)
```

Bài 27: Cho một dãy số nguyên, Tìm và in ra vị trí các số < x (x là giá trị cho trước).
nếu không có in ra -1


```python
a = list(map(int, input().split()))
x = int(input())

found = False

for i in range(len(a)):
    if a[i] < x:
        print(i, end=" ")
        found = True

if not found:
    print(-1)
```

Bài 28: Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm
thực sự, hay vô danh (không tăng, không giảm)

```python
a = list(map(int, input().split()))

tang = True
giam = True

for i in range(len(a) - 1):
    if a[i] >= a[i + 1]:
        tang = False

    if a[i] <= a[i + 1]:
        giam = False

if tang:
    print("Tang thuc su")
elif giam:
    print("Giam thuc su")
else:
    print("Vo danh")
```
